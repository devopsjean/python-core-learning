# Loops

Use `for` when iterating over a sequence and `while` when repeating until a condition changes.

## Step 1
Run a `for` loop with `range()`. Notice where the sequence starts and where it stops.

```python
print("for loop")
for number in range(1, 6):
    print(number)
```

## Step 2
Run a `while` loop and update the condition inside the loop. This prevents the loop from running forever.

```python
print("while loop")
countdown = 3
while countdown > 0:
    print(countdown)
    countdown -= 1

print("done")
```
