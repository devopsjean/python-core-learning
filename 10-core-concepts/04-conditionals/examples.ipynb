# Conditionals

See how `if`, `elif`, and `else` change the program path.

## Step 1
Set up values that control the branch logic. Then read the first condition block and predict the output before running it.

```python
temperature = 28
is_raining = False

if temperature >= 30:
    print("hot")
elif temperature >= 20:
    print("warm")
else:
    print("cold")
```

## Step 2
Combine multiple conditions with `not` and `and`. This shows how simple boolean expressions shape a decision.

```python
if not is_raining and temperature >= 20:
    print("good for a short walk")
else:
    print("stay inside and study")
```
