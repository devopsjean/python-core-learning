# Errors and Exceptions

Catch runtime problems and keep the program under control.

## Step 1
Define two functions that protect risky operations with `try` and `except`. The goal is not to hide errors, but to handle them safely.

```python
def safe_divide(left, right):
    try:
        return left / right
    except ZeroDivisionError:
        return "cannot divide by zero"


def parse_int(text):
    try:
        return int(text)
    except ValueError:
        return "invalid integer"
```

## Step 2
Run both functions with valid and invalid input. Compare normal results with handled error cases.

```python
print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(parse_int("123"))
print(parse_int("abc"))
```
