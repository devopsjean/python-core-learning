# Collections

Compare list, tuple, dictionary, and set in one place.

## Step 1
Create one example of each collection type. The goal is to see how each structure stores multiple values differently.

```python
services = ["api", "worker", "scheduler"]
coordinates = (37.5, 127.0)
status_codes = {"ok": 200, "not_found": 404}
ports = {80, 443, 443}

services.append("batch")
```

## Step 2
Read values back from each collection. Watch how indexing, keys, and membership checks differ.

```python
print(services)
print(services[0])
print(coordinates)
print(status_codes["ok"])
print(sorted(ports))
print("worker" in services)
```
