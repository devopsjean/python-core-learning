# Input and Output

This notebook uses simulated input values so it runs cleanly in GitHub and Jupyter.

## Step 1
Define a small function that builds a message. This keeps the output logic readable and reusable.

```python
def build_greeting(name, topic):
    return f"Hello, {name}. Today you study {topic}."
```

## Step 2
Use fixed values as simulated input. This avoids interactive input while still showing the same idea.

```python
simulated_name = "Junsung"
raw_score = "85"
score = int(raw_score)

print("Study session started")
print(build_greeting(simulated_name, "conditionals"))
print(f"Score plus five: {score + 5}")
```
