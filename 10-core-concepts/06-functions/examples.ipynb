# Functions

Define reusable logic and call it with input values.

## Step 1
Define three small functions. Each one focuses on a single job: greeting, adding, or checking a condition.

```python
def greet(name):
    return f"Hello, {name}"


def add(left, right):
    return left + right


def is_even(number):
    return number % 2 == 0
```

## Step 2
Call the functions and inspect the returned values. This is where input and output become visible.

```python
message = greet("Junsung")
total = add(10, 5)

print(message)
print(total)
print(is_even(6))
print(is_even(7))
```
