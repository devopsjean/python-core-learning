# Modules

Import standard library modules and use them directly.

## Step 1
Import modules from the standard library. This shows that Python code can be organized and reused across files.

```python
import math
import random
from pathlib import Path
```

## Step 2
Call functions and objects from imported modules. This is the practical side of using modules in everyday code.

```python
random.seed(7)

print(math.sqrt(49))
print(Path.cwd().name)
print(random.choice(["variables", "loops", "functions"]))
```
